```python
import pandas as pd
df=pd.read_csv("C:/Users/User/Downloads/Telegram Desktop/cars2018.csv")
from sklearn.preprocessing import LabelEncoder , MinMaxScaler
import numpy as np
class DataPreprocessing:
    def __init__(self, df):
        self.df=df.copy()
    def encoding(self):
        encoder = LabelEncoder()
        for col in self.df.columns:
            if self.df[col].dtype == "object":
                if self.df[col].nunique() <= 5:
                    dummies = pd.get_dummies(self.df[col], dtype=int, prefix=col)
                    self.df = pd.concat((self.df.drop(columns=col), dummies), axis=1)
                else:
                    self.df[col] = encoder.fit_transform(self.df[col])
        return self
    
    def scaling(self):
        scaler=MinMaxScaler()
        num_col=self.df.select_dtypes(include=["float64","int64"]).columns.drop("mpg")
        self.df[num_col]=scaler.fit_transform(self.df[num_col])
        return self
    def log_transform(self, threshold=0.5):
        skewness = self.df.skew()
        features_log = skewness[(skewness >= threshold)].index.tolist()
        
        for col in features_log:
            if (self.df[col] > 0).all():  # only positive values
                self.df[col] = np.log1p(self.df[col])
        return self

dp=DataPreprocessing(df)
dp.encoding()
dp.scaling().log_transform()
df=dp.df
```
